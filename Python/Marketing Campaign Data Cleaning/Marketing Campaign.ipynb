```python
import pandas as pd
import numpy as np
df = pd.read_csv(r"C:\Etudes\projects\Python\Marketing Compaign\marketing_campaign_data_messy.csv",
                 parse_dates=["Start_Date","End_Date"])
df
```

```python
df.info()
```

```python
# Cleaning Header
print(df.columns.to_list())
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
print("Fix Applied:")
print(df.columns.to_list())
```

```python
# Type Conversion and Currency Cleaning
df["spend"] = df["spend"].replace(r"[^\d.]", "", regex=True).astype("float")
df.info()
```

```python
# Categorical Typos
print(df["channel"].unique())
cleanup_map = {
    "Facebok": "Facebook",
    "Gogle": "Google Ads",
    "Tik_Tok": "TikTok",
    "E-mail": "Email",
    "Insta_gram": "Instagram",
    "N/A": np.nan
}
df["channel"] = df["channel"].replace(cleanup_map)

print("Fix Applied:")
print(df["channel"].unique())
```

```python
# Handling Mixed Booleans
print(df["active"].unique())

boolean_map = {
    "True": True,
    "Yes": True,
    "1": True,
    "Y": True,
    "0": False,
    "No": False
}
df["active"] = df["active"].replace(boolean_map).fillna(False).astype("bool")
print("Fix Applied:")
print(df["active"].unique())
```

```python
# Dates Parsing
print(df["start_date"].dtype)

df["start_date"] = pd.to_datetime(df["start_date"], errors="coerce")
print("Fix Applied:")
print(df["start_date"].dtype)
```

```python
df = df.loc[:, ~df.columns.duplicated()]
```

```python
# Logical Integrity(clicks vs impressions)
impossible_mask = df["impressions"] < df["clicks"]
print(df.loc[impossible_mask, ["clicks", "impressions"]].head())
```

```python
# Logical Integrity(time travel)
time_travel_mask = df["start_date"] > df["end_date"]
print(len(df.loc[time_travel_mask, ["start_date", "end_date"]]))
df["days_diff"] = (df["end_date"] - df["start_date"]).dt.days
correct_average = df.loc[df["days_diff"] >= 0, "days_diff"].mean()
wrong_average = df.loc[df["days_diff"] < 0, "days_diff"].mean()
print(f" Average days difference of correct values: {correct_average}")
print(f" Average days difference of wrong values: {wrong_average}")

# Assumption: in time travel_travel_mask start_date and end_date were inversed
df.loc[time_travel_mask, ["start_date", "end_date"]] = df.loc[time_travel_mask, ["end_date", "start_date"]].values
print("Fix Applied:")
print(df.loc[time_travel_mask, ["start_date", "end_date"]].head())
df.drop(columns=["days_diff"], inplace=True)
```

```python
# Handling Outliers

Q1 = df["spend"].quantile(0.25)
Q3 = df["spend"].quantile(0.75)

IQR = Q3 - Q1
upper_limit = Q3 + (3*IQR)

outlier_mask = df["spend"] > upper_limit
print(df.loc[outlier_mask, "spend"].head(3))

df.loc[outlier_mask, "spend"] =  upper_limit
print("Fix Applied:")
print(df.loc[outlier_mask, "spend"].head(3))
```

```python
# String Parsing

print(df["campaign_name"].head(3))

df["season"] = df["campaign_name"].str.split("_").str[1]
print("Fix Applied:")
print(df[["campaign_name", "season"]].head(3))
```
