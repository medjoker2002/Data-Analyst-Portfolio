```python
import pandas as pd
from pandas import DataFrame

df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Making The Cut\marathon-data.csv")
df.head()
```

```python
from numpy import select
count_all = len(df)

# df["final"] = pd.to_datetime(df["final"], format="%H:%M:%S")
conditions =[
    df["final"] < "03:00:00",
    df["final"] < "03:30:00",
    df["final"] < "04:00:00",
    df["final"] < "04:30:00",
    df["final"] < "05:00:00",
    df["final"] < "05:30:00"
]
choices = [
    "sub 3:00:00",
    "3:00:00-3:30:00",
    "3:30:00-4:00:00",
    "4:00:00-4:30:00",
    "4:30:00-5:00:00",
    "5:00:00-5:30:00"
]
df["finish_band"] = select(conditions, choices, default="6:00:00+")
df.head()
```

```python
df_count = (
    df.groupby("finish_band")
    .agg(runners = ('gender', "count"))
)
df_count["% of field"] = (df_count["runners"] * 100/ count_all).round(2)
df_count.sort_values(by= "finish_band")
df_count
```
