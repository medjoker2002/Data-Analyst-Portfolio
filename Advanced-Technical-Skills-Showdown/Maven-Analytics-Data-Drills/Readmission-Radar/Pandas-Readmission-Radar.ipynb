```python
import pandas as pd
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Readmission Radar\inpatient_admissions.csv",
                 parse_dates=['admission_date', 'discharge_date'])
df.head()
```

```python
df["readmission_date"] = (
    df.sort_values(by="admission_date")
    .groupby("patient_id")["admission_date"]
    .shift(-1)
)
df["duration"] = (df["readmission_date"] - df["discharge_date"]).dt.days
df.head()
```

```python
from numpy import where
df["readmitted_in_30_days"] = where(df["duration"] <= 30, 1, 0)
df.head()
```

```python
readmission_rate = df.loc[:, "readmitted_in_30_days"].sum() / len(df)
print(f"Readmission rate: {readmission_rate}")
```
