```python
import pandas as pd
df_users = pd.read_csv(r"C:\Etudes\projects\data drills maven\Movie Metrics\Users.csv")
df_acts = pd.read_csv(r"C:\Etudes\projects\data drills maven\Movie Metrics\activity.csv", parse_dates=["date"])
df_acts.head()
```

```python
df_fin = df_acts[df_acts["finished"] == 1]
df_fin.head(10)
```

```python
df_started = df_acts.groupby(by = "user_id").agg({"id": "count"})
df_started.rename(columns= {"id": "started"}, inplace = True)
df_started
```

```python
df_metrics = (
    df_fin
    .sort_values(by = ["user_id", "date", "id"])
    .groupby(by= "user_id")
    .agg(
        {"user_id": "count",
         "date": ["min", "max"],
         "movie_name": ["first", "last"],})
)
df_metrics.columns = ["finished", "first_date", "last_date", "first_movie", "last_movie"]
df_metrics
```

```python
df_metrics = df_started.merge(df_metrics, on = "user_id", how = "inner")
df_metrics
```
