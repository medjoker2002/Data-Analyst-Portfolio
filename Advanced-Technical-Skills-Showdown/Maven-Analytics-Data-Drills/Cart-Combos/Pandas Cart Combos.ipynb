```python
import pandas as pd

df = pd.read_csv(
    r"C:\Etudes\projects\data drills maven\Cart Combos\grocery_transactions.csv",
    usecols=["transaction_id", "product_name"]
)
df.head()
```

```python
df["product_id"] = df["product_name"].rank(method="dense").astype(int)
df.head()
```

```python
df_merged = pd.merge(df, df, on="transaction_id", how="inner", suffixes=("_1", "_2"))
df_merged = df_merged.query("product_id_1 < product_id_2")
df_merged = df_merged.reset_index(drop=True)
df_merged.head()
```

```python
df_pairs = (
    df_merged
    .groupby(["product_name_1", "product_name_2"])
    .agg(number_of_transactions = ("transaction_id", "count"))
)
df_pairs.sort_values(by="number_of_transactions", ascending=False, inplace=True)
df_pairs
```
