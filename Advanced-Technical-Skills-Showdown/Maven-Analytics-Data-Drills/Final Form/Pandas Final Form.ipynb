```python
import pandas as pd
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Final Form\employee_satisfaction_survey.csv", parse_dates=["Timestamp"])
df.head(10)
```

```python
df["Timestamp"] = pd.to_datetime(df["Timestamp"])
df_last = (
    df.sort_values(by="Timestamp")
    .groupby("Email")
    .agg({"Satisfaction": "last"})
)
df_last.head(10)
```

```python
df_count = (
    df_last.groupby("Satisfaction")
    .agg(Employees = ("Satisfaction", "count"))
)
df_count.sort_values(by="Satisfaction", ascending=False)
df_count
```
