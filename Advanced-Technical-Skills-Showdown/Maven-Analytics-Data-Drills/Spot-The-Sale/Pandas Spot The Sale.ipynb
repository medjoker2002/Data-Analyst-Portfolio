```python
import pandas as pd
import numpy as np
df_orders = pd.read_csv(r"C:\Etudes\projects\data drills maven\Spot The Sale\orders.csv", parse_dates=["order_date"])
df_promotions = pd.read_csv(r"C:\Etudes\projects\data drills maven\Spot The Sale\promotions.csv", parse_dates=["start_date", "end_date"])
df_promotions.head(30)
```

```python
merged = pd.merge_asof(
    df_orders.sort_values("order_date").reset_index(drop=True),
    df_promotions.sort_values("start_date").reset_index(drop=True),
    left_on="order_date",
    right_on="start_date",
    direction="backward"
)
merged = merged.query("order_date <= end_date")
merged
```
