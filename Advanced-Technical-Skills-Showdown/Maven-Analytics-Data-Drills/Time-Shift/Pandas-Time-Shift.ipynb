```python
import pandas as pd
df_users = pd.read_csv(r"C:\Etudes\projects\data drills maven\Time Shift\users.csv")
df_tickets = pd.read_csv(r"C:\Etudes\projects\data drills maven\Time Shift\tickets.csv")
df_tickets["submitted_at_utc"] = pd.to_datetime(df_tickets["submitted_at_utc"].str.replace("T", " "))
df_users.head()
```

```python
df_users["timezone"] = df_users["timezone"].apply(lambda x: str(x)[4:7]).astype("int32")
df_users.head()
```

```python
df_tickets = df_tickets.merge(df_users, on="user_id", how="left")
df_tickets.head()
```

```python
df_tickets["local_time"] = df_tickets["submitted_at_utc"] + pd.to_timedelta(df_tickets["timezone"], unit="h")
df_tickets.head()
```

```python
df_hourly =(
    df_tickets
    .groupby(df_tickets["local_time"].dt.hour)
    .agg(count_tickets = ("ticket_id", "count"))
)
df_hourly.sort_values(by="count_tickets", ascending=False, inplace=True)
df_hourly
```
