```python
import pandas as pd
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Golden cross\SPY_close_price_5Y.csv", parse_dates=["Date"])
df.head(30)
```

```python
df["ma50"] = df["Close"].rolling(50).mean()
df["ma200"] = df["Close"].rolling(200).mean()
df.head(60)
```

```python
df["lag_ma50"] = df["ma50"].shift(1)
df["lag_ma200"] = df["ma200"].shift(1)
df.head(60)
```

```python
from numpy import where
df["golden_cross"] = where(
    (df["ma50"] > df["ma200"]) * (df["lag_ma50"] <= df["lag_ma200"]),
    1,
    0
)
df
```
