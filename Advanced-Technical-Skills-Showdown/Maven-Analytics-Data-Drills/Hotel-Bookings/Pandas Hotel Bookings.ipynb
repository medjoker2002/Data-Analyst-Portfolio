```python
import pandas as pd
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Hotel Bookings\hotel_bookings.csv", parse_dates=["booking_date", "cancel_date", "checkin_date", "checkout_date"])
df.head(30)
```

```python
df = df.query("is_canceled==0")
df.drop(columns=["is_canceled", "cancel_date", "booking_id", "booking_date"], inplace=True)
df.head(30)
```

```python
print(f"start: {min(df['checkin_date'])}")
print(f"end: {max(df['checkout_date'])}")
```

```python
dates = pd.date_range(start="2015-07-01", end="2017-09-13")
df_dates = pd.DataFrame(dates, columns=["date"])
df_dates.head(20)
```

```python
df_dates["rooms_available"] = 200
df_dates["occupied_rooms"] = df_dates["date"].apply(
    lambda x: ((df["checkin_date"] <= x) & (x < df["checkout_date"])).sum()
)
df_dates.head(30)
```

```python
df_dates["yy-mm"] = df_dates["date"].apply(lambda x: str(x)[ :7])
df_dates["occupancy_rate"] = (df_dates["occupied_rooms"] * 100 / df_dates["rooms_available"]).round(2)
df_dates.head(30)
```

```python
df_monthly = (
    df_dates
    .groupby("yy-mm")
    .agg({"occupancy_rate": "mean"}).round(2)
)
df_monthly
```
