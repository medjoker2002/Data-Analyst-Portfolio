```python
import pandas as pd
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Org Chart Overhaul\OfficeSpace.csv")
df
```

```python
df["reporting_hierarchy"] = ""
for i in df.index:
    last_manager = df.loc[i,"Manager Name"]
    managers = [df.loc[i,"Employee Name"], last_manager]
    while not pd.isna(last_manager):
        for j in df.index:
            if last_manager == df.loc[j,"Employee Name"]:
                last_manager = df.loc[j,"Manager Name"]
                managers.append(last_manager)
                break
    df.loc[i, "reporting_hierarchy"] = managers[-2]
    for m in range(-3, -len(managers)-1, -1):
        df.loc[i, "reporting_hierarchy"] += " > " + managers[m]
df
```

```python
df["direct_reports"] = df["Employee Name"].apply(lambda x: (df["Manager Name"] == x).sum())
df
```

```python
df["total_reports"] = df["reporting_hierarchy"].apply(lambda x: (df["reporting_hierarchy"].str.startswith(x)).sum()) - 1
df
```
