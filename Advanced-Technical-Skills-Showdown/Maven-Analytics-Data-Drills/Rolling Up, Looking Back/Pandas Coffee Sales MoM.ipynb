```python
import pandas as pd
import numpy as np
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Coffee Sales MoM\coffee_shop_sales.csv", parse_dates=["date"])
```

```python
df.head(20)
```

```python
df["month"] = df["date"].dt.month
df_monthly = (
    df.groupby(["month", "store"])
    .agg({"sales": "sum"})
)
df_monthly
```

```python
df_monthly["prv_month"] = df_monthly.groupby("store")["sales"].shift(1)
df_monthly["MoM"] = df_monthly["sales"] - df_monthly["prv_month"]
df_monthly
```
