```python
import numpy as np
import pandas as pd
df_estate = pd.read_csv(r"C:\Etudes\projects\data drills maven\Estimate The Estate\manhattan_property_sales.csv")
df_estate.head(20)
```

```python
df_estate["SALE_PRICE"] = np.where(df_estate["SALE_PRICE"]>0, df_estate["SALE_PRICE"], np.nan)
df_estate["price/sqrft"] = df_estate["SALE_PRICE"] / df_estate["SQUARE_FEET"]
df_estate["avg_price/sqrft"] = (
    df_estate
    .groupby(["ZIP_CODE", "BUILDING_CLASS"])["price/sqrft"]
    .transform("mean")
)
df_estate.head(20)
```

```python
df_estate["market_value"] = np.where(df_estate["SALE_PRICE"]>0,
                                     df_estate["SALE_PRICE"],
                                     df_estate["avg_price/sqrft"]*df_estate["SQUARE_FEET"])
df_estate.head(20)
```

```python
# number of estates with a market value over 15M
print(len(df_estate[df_estate["market_value"]>15000000]))
```
