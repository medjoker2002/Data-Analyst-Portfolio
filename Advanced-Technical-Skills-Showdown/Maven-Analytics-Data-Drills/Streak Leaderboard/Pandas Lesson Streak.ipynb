```python
import pandas as pd
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\Lesson Streak\LessonStreaks.csv", parse_dates=["date"])
df.head(30)
```

```python
df.drop(columns=["rn"], inplace=True)
df.drop_duplicates(ignore_index=True, inplace=True)
df["rn"] = (
    df.sort_values(by="date", ascending=True)
    .groupby("user_id")
    .cumcount() + 1
)
df.sort_values(by= ["user_id", "date"]).head(30)
```

```python
df['streak_grp'] = df["date"] - pd.to_timedelta(df["rn"], unit= "D")
df.head(30)
```

```python
df_streaks = (
    df.groupby(["user_id", "streak_grp"])
    .agg({"user_name": "max" ,"date": ["min", "max"], "user_id": "count"})
)
df_streaks.columns = ["user_name", "start_date", "end_date", "streak_length"]
df_streaks
```

```python
active_streaks = df_streaks[df_streaks["end_date"]== "2025-09-28"].sort_values(by="streak_length", ascending=False)
active_streaks
```
