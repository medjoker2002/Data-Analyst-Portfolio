```python
from shlex import split

import pandas as pd
df = pd.read_csv(r"C:\Etudes\projects\data drills maven\2025 MLB positions\baseball_positions.csv")
df.head(20)
```

```python
positions = []
for i in df["Position"]:
    positions += i.split("/")
```

```python
df_positions = pd.DataFrame.from_dict(
    {"Position": positions}
)
df_positions
```

```python
df_positions = (
    df_positions
    .groupby("Position")
    .agg({"Position": "count"})
)
df_positions.columns = ["Count"]
df_positions.sort_values(by="Count", ascending=False, inplace=True)
df_positions
```
